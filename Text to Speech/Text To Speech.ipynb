```python
import pyttsx3
import threading

tts = None

def speaknow():
    global tts

    text = text_box.get("1.0", END).strip()

    if not text:
        return

    gender = gender_box.get()
    speed = speed_box.get()

    def speak():
        global tts

        tts = pyttsx3.init()

        voices = tts.getProperty("voices")

        # Select voice
        if gender == "Male":
            tts.setProperty("voice", voices[0].id)
        else:
            tts.setProperty("voice", voices[1].id)

        # Select speed
        if speed == "Fast":
            tts.setProperty("rate", 250)
        elif speed == "Medium":
            tts.setProperty("rate", 150)
        else:
            tts.setProperty("rate", 60)

        tts.setProperty("volume", 1.0)

        tts.say(text)
        tts.runAndWait()

        tts.stop()
        tts = None

    # Run speech separately
    threading.Thread(target=speak, daemon=True).start()


def stop_speech():
    global tts

    if tts is not None:
        tts.stop()
        tts = None
```

```python
from tkinter import *
from PIL import Image, ImageTk

root = Tk()
root.title("Text to Speech converter")
root.geometry("1000x600+200+80")
root.resizable(False, False)
root.configure(bg="#F7AC40")


# Logo in title bar
logo_image = PhotoImage(file=r"E:\All My Projects\AIML\logo.png")
root.iconphoto(False, logo_image)


# Upper blue frame
upper_frame = Frame(
    root,
    bg="#14A7DD",
    width=1000,
    height=130
)
upper_frame.place(x=0, y=0)


# Logo inside the blue frame
img = Image.open(r"E:\All My Projects\AIML\logo.png")
img.thumbnail((280, 120))

picture = ImageTk.PhotoImage(img)

Label(
    upper_frame,
    image=picture,
    bg="#14A7DD"
).place(x=150, y=10)


#root.mainloop()
```

```python
text_box = Text(root, font="calibri 20", bg ="white", relief = GROOVE, wrap= WORD, bd=0)
text_box.place(x=30, y=150, width =940, height=180)
#root.mainloop()
```

```python
from PIL import Image, ImageTk

# Load Play image using Pillow
play_img = Image.open(r"E:\All My Projects\AIML\Play.png")

# Resize it
play_img = play_img.resize((40, 40))

# Convert to Tkinter image
play_button = ImageTk.PhotoImage(play_img)

# Create button
play_btn = Button(
    root,
    text="Play",
    compound=LEFT,
    image=play_button,
    bg="white",
    width=130,
    font="arial 14 bold",
    borderwidth=0,
    command=speaknow
)

play_btn.place(x=350, y=490)

#root.mainloop()
```
